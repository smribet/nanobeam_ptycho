# iterative ptycho

```python
from quantem.core.datastructures import Dataset4dstem
from quantem.diffractive_imaging.dataset_models import PtychographyDatasetRaster
from quantem.diffractive_imaging.ptychography_lite import PtychoLite, PtychoLiteDIP
```

```python
pdset = PtychographyDatasetRaster.from_dataset4dstem(dc)

pdset.preprocess(
    com_fit_function="constant",
    plot_rotation=True,
    plot_com=True,
    probe_energy=energy,
    force_com_rotation=-169,
    force_com_transpose = False,
)
```

```python
import os
import random
import numpy as np
import torch
import h5py
import gc
import torch 

torch.cuda.empty_cache()
gc.collect()


os.makedirs("recons", exist_ok=True)

seeds = [67, 72, 92]

constraints = {
    "object": {
        "identical_slices": True,
    },
    "probe": {
        "orthogonalize_probe": False,
    },
}

for run_idx, seed in enumerate(seeds):
    print(f"Running reconstruction {run_idx + 1}/3 with seed {seed}")
    torch.cuda.empty_cache()
    gc.collect()
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    ptycho_pix = PtychoLite.from_dataset(
        dset=pdset,
        num_slices=1,
        obj_type="complex",
        num_probes=4,
        energy=energy,
        semiangle_cutoff=probe_semiangle,
        obj_padding_px=(32, 32),
        device="cpu",
        polar_parameters={
            'C10':  -176.7006242519572,
        },
    )

    ptycho_pix.reconstruct(
        num_iters=50,
        device=[0, 1, 2],
        batch_size=512 * 3,
        constraints=constraints,
        reset=True,
        lr_obj=5e-3,
        lr_probe=5e-2,
    ).visualize()

    out_path = f"recons/260706_scan_23_pix_ptycho_seed_{seed}.h5"
    with h5py.File(out_path, "w") as f:
        f.attrs["seed"] = seed
        f.attrs["run_idx"] = run_idx
        f.create_dataset("obj", data=ptycho_pix.obj_cropped)
        f.create_dataset("sampling", data=ptycho_pix.sampling)
        f.create_dataset("probe", data=ptycho_pix.probe)

    print(f"Saved {out_path}")
```
